```python
import pandas as pd
from sklearn.neural_network import MLPClassifier
from sklearn.model_selection import train_test_split

# Read in data from files output by Stemming.ipynb
# * Stemming.ipynb will need to be run first to generate these files
X_stemmed = pd.read_csv("data/X_train_counts.csv")
y_stemmed = pd.read_csv("data/y_train.csv")
```

```python
# Split the data into training and validation sets
X_train, X_val, y_train, y_val = train_test_split(X_stemmed, y_stemmed, test_size=.2, random_state=1)
print("X_train Shape:", X_train.shape)
print("y_train Shape:", y_train.shape)
```

```python
# Initialize default-ish classifier
hidden_layer_size = 100                 # Number of neurons in each hidden layer (default is 100, ) <- 1 hidden layer with 100 neurons
activation = 'relu'                     # No idea what this means, but only has 4 options: 'identity', 'logistic, 'tanh', 'relu' (default is 'relu')
solver = 'adam'                         # Weight optimization algorithm: only has 4 options: 'lbfgs', 'sgd', 'adam', 'ibfgs' (default is 'adam')
alpha = 0.0001                          # Regularization term: increase to fix overfitting and decrease to fix underfitting (default is 0.0001)
batch_size = 'auto'                     # Number of samples per batch !Not used by 'ibfgs'! (default is 'auto', which is min(200, n_samples))
learning_rate = 'constant'              # How/if learning rate varies. Only 3 options: 'constant', 'invscaling', 'adaptive' !only used by 'sgd'! (default is 'constant's)
learning_rate_init = 0.001              # Initial learning rate !Not used by 'ibfgs'! (default is 0.001)
power_t = 0.5                           # Exponent for inverse scaling learning rate !only used by 'sgd' when 'inviscaling'! (default is 0.5)
max_iter = 200                          # Maximum number of iterations (default is 200)
shuffle = True                          # !Just keep true! Whether to shuffle samples in each iteration (default is True)
random_state = 1                        # !Just keep 1 for hyperparameter optimization! Random seed for weight initialization (default is None)
tol = 1e-4                              # !Probably keep and 1e-4! Tolerance for optimization !Not used when 'adaptive'! (default is 1e-4)
verbose = False                         # !Just keep false! Whether to print progress messages to stdout (default is False)
warm_start = False                      # Do some research
momentum = 0.9                          # Momentum for gradient descent update !only used by 'sgd'! (default is 0.9)
nesterovs_momentum = True               # Whether to use Nesterov's momentum !only used by 'sgd'! (default is True)
early_stopping = False                  # !Just keep false! Whether to stop training when validation score is not improving !only used by 'sgd' and 'adam'! (default is False)
validation_fraction = 0.1               # !Used by early_stopping! Proportion of training data to set aside as validation set (default is 0.1)
beta_1 = 0.9                            # Exponential decay rate for estimates of first moment vector in adam !only used by 'adam'! (default is 0.9)
beta_2 = 0.999                          # Exponential decay rate for estimates of second moment vector in adam !only used by 'adam'! (default is 0.999)
epsilon = 1e-8                          # Value for numerical stability in adam !only used by 'adam'! (default is 1e-8)
n_iter_no_change = 10                   # !Probably keep 10! Maximum number of epochs to not meet tol improvement !only used by 'sgd' and 'adam'! (default is 10)

clf = MLPClassifier(
    hidden_layer_sizes=hidden_layer_size,
    activation=activation,
    solver=solver,
    alpha=alpha,
    batch_size=batch_size,
    learning_rate=learning_rate,
    learning_rate_init=learning_rate_init,
    power_t=power_t,
    max_iter=max_iter,
    shuffle=shuffle,
    random_state=random_state,
    tol=tol,
    verbose=verbose,
    warm_start=warm_start,
    momentum=momentum,
    nesterovs_momentum=nesterovs_momentum,
    early_stopping=early_stopping,
    validation_fraction=validation_fraction,
    beta_1=beta_1,
    beta_2=beta_2,
    epsilon=epsilon,
    n_iter_no_change=n_iter_no_change
)
clf.fit(X_train, y_train.values.ravel())
```

```python
# Evaluate the classifier
clf.score(X_val, y_val.values.ravel())
#clf.n_layers_
```
